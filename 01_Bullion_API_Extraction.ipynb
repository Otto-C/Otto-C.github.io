```python
from google.colab import userdata

api_key = userdata.get('METALS_API_KEY')

print("API key loaded successfully!" if api_key else "API key not found")
```

```python
import requests
url = 'https://api.metals.dev/v1/latest'

params = {
    "api_key": api_key,
    "currency": "ZAR",
    "unit": "toz"
}

response = requests.get(url, params=params, timeout=30)

print("HTTP Status", response.status_code)
print("API Status", response.json().get("status"))
```

```python
data = response.json()
print(data)
```

```python
import json

print(json.dumps(data, indent=4))
```

```python
gold_price = data["metals"]["gold"]
silver_price = data["metals"]["silver"]
platinum_price = data["metals"]["platinum"]
palladium_price = data["metals"]["palladium"]
status_code = data["status"]
currency = data["currency"]
zar = data["unit"]


```

```python
print(gold_price)
print(silver_price)
print(status_code)
print(currency)
print(zar)
print("")
print("Gold Price", round(gold_price,2))
print("Silver Price", round(silver_price, 2))
print("Platinum Price", round(platinum_price, 2))
print("Palladium Price", round(palladium_price, 2))

print("")
print("Status Code", status_code)
print("Currency", currency)
print("ZAR", zar)
```
